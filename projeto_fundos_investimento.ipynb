```python
import pandas as pd

# Carregar o dataset da CVM
df = pd.read_csv('dataset/inf_diario_fi_202609.csv', sep=';')

# Exibir as primeiras 5 linhas da tabela
df.head()
```

```python
# Verificar a quantidade de linhas e colunas
print("Formato do dataset (Linhas, Colunas):", df.shape)

# Exibir informações das colunas e tipos de dados
df.info()
```

```python
# Converter a coluna de data para o formato datetime
df['DT_COMPTC'] = pd.to_datetime(df['DT_COMPTC'])

# Verificar o intervalo de datas do arquivo
print("Data inicial:", df['DT_COMPTC'].min())
print("Data final:", df['DT_COMPTC'].max())

# Estatísticas descritivas do Patrimônio Líquido
df[['VL_PATRIM_LIQ', 'VL_QUOTA']].describe()
```

```python
# Formatar visualização do Pandas para mostrar decimais normais sem notação científica
pd.options.display.float_format = '{:,.2f}'.format

# Filtrar os dados do último dia disponível (28/09/2026)
ultimo_dia = df[df['DT_COMPTC'] == '2026-09-28']

# Listar os 10 maiores fundos por Patrimônio Líquido nesse dia
maiores_fundos = ultimo_dia.sort_values(by='VL_PATRIM_LIQ', ascending=False).head(10)

# Exibir CNPJ e Patrimônio Líquido dos maiores fundos
maiores_fundos[['CNPJ_FUNDO_CLASSE', 'VL_PATRIM_LIQ']]
```

```python
# PERGUNTA 1: Quantos fundos diferentes existem no dataset?
qtd_fundos = df['CNPJ_FUNDO_CLASSE'].nunique()
print("PERGUNTA 1: Quantos fundos diferentes existem no dataset?")
print(f"Resultado: {qtd_fundos:,} fundos distintos")

print("\n💡 ANÁLISE PARA O GERENTE:")
print(f"O mercado analisado apresenta uma elevada diversidade de veículos, com {qtd_fundos:,} opções ativas de fundos.")
print("Orientação ao cliente: Diante de tantas opções, o papel do gerente é atuar como curador. ")
print("Em vez de expor o cliente a excesso de escolhas, selecione opções alinhadas estritamente ao seu perfil e horizonte de investimento.")
```

```python
# PERGUNTA 2: Qual é o patrimônio líquido médio dos fundos no período?
pl_medio = df['VL_PATRIM_LIQ'].mean()
print("PERGUNTA 2: Qual é o patrimônio líquido médio dos fundos no período?")
print(f"Resultado: R$ {pl_medio:,.2f}")

print("\n💡 ANÁLISE PARA O GERENTE:")
print(f"O PL médio de R$ {pl_medio:,.2f} serve como referência para avaliar o porte dos produtos oferecidos no portfólio.")
print("Orientação ao cliente: Fundos que operam muito abaixo desta média de mercado exigem atenção à taxa de administração e ao risco de descontinuidade (fusão ou liquidação).")
print("Recomende ao cliente a preferência por fundos consolidados.")
```

```python
# PERGUNTA 3: Qual é o maior e o menor patrimônio líquido registrado?
pl_max = df['VL_PATRIM_LIQ'].max()
pl_min = df['VL_PATRIM_LIQ'].min()
print("PERGUNTA 3: Qual é o maior e o menor patrimônio líquido registrado?")
print(f"Resultado - Maior PL: R$ {pl_max:,.2f}")
print(f"Resultado - Menor PL: R$ {pl_min:,.2f}")

print("\n💡 ANÁLISE PARA O GERENTE:")
print("A enorme discrepância entre o maior e o menor PL evidencia a coexistência de grandes fundos institucionais e fundos pequenos ou em reestruturação.")
print("Orientação ao cliente: Explique ao cliente que fundos gigantes têm maior resiliência e poder de negociação de ativos, enquanto fundos muito pequenos só se justificam se apresentarem uma estratégia de nicho comprovadamente diferenciada.")
```

```python
# PERGUNTA 4: Qual é o número médio de cotistas dos fundos?
media_cotistas = df['NR_COTST'].mean()
print("PERGUNTA 4: Qual é o número médio de cotistas dos fundos?")
print(f"Resultado: {media_cotistas:,.0f} cotistas em média")

print("\n💡 ANÁLISE PARA O GERENTE:")
print(f"A média de {media_cotistas:,.0f} cotistas indica a presença relevante de fundos voltados ao público de varejo pulverizado.")
print("Orientação ao cliente: Quanto maior a base de cotistas, menor é a volatilidade causada por saques individuais. Destaque para o cliente que fundos bem pulverizados oferecem maior previsibilidade de gestão no dia a dia.")
```

```python
# PERGUNTA 5: Quais são os 5 fundos com mais de 1.000 cotistas no último dia disponível?
filtro_cotistas = df[(df['DT_COMPTC'] == '2026-09-28') & (df['NR_COTST'] > 1000)]
top5_pulverizados = filtro_cotistas[['CNPJ_FUNDO_CLASSE', 'NR_COTST', 'VL_PATRIM_LIQ']].sort_values(by='NR_COTST', ascending=False).head(5)
print("PERGUNTA 5: Quais são os 5 fundos com mais de 1.000 cotistas no último dia disponível?")
print("Resultado - Top 5 Fundos Pulverizados:")
display(top5_pulverizados)

print("\n💡 ANÁLISE PARA O GERENTE:")
print("Este grupo representa os fundos mais populares e acessíveis do mercado, com altíssima pulverização de investidores.")
print("Orientação ao cliente: Ideal para apresentação a clientes de varejo e novos investidores que valorizam a validação do mercado e a segurança de estar em veículos amplamente testados pelo público.")
```

```python
# PERGUNTA 6: Qual categoria de fundo apresenta o maior patrimônio líquido médio?

print("PERGUNTA 6: Qual categoria de fundo apresenta o maior patrimônio líquido médio?")


pl_por_classe = df.groupby('TP_FUNDO_CLASSE')['VL_PATRIM_LIQ'].mean().sort_values(ascending=False)

print("Resultado - PL Médio por Categoria:")
display(pl_por_classe.to_frame())

print("\n💡 ANÁLISE PARA O GERENTE:")
print("- O maior PL médio indica onde o mercado concentra os maiores volumes institucionais.")
print("- Orientação ao cliente: Categorias com PL médio muito elevado costumam oferecer maior estabilidade e menor risco de liquidação do fundo. São ideais para alocação do 'core' da carteira de investidores conservadores ou moderados.")
```

```python
# PERGUNTA 7: Existe diferença no PL médio entre fundos com menos e mais de 1.000 cotistas?

print("PERGUNTA 7: Existe diferença no PL médio entre fundos com < 1.000 e >= 1.000 cotistas?")


pl_pequenos = df[df['NR_COTST'] < 1000]['VL_PATRIM_LIQ'].mean()
pl_grandes = df[df['NR_COTST'] >= 1000]['VL_PATRIM_LIQ'].mean()

print(f"Resultado - PL Médio (< 1.000 cotistas):  R$ {pl_pequenos:,.2f}")
print(f"Resultado - PL Médio (>= 1.000 cotistas): R$ {pl_grandes:,.2f}")

print("\n💡 ANÁLISE PARA O GERENTE:")
print("- Fundos pulverizados (>= 1.000 cotistas) tendem a ter menor risco de iliquidez provocado por um único cotista.")
print("- Orientação ao cliente: Se o cliente tem perfil de varejo e busca liquidez rápida (D+0 a D+5), fundos com grande base de cotistas reduzem o impacto de saques repentinos na rentabilidade do produto.")
```

```python
# PERGUNTA 8: Qual categoria apresenta a maior média de captação e resgate diários?

print("PERGUNTA 8: Qual categoria apresenta a maior média de captação e resgate por dia?")


resumo_capt_resg = df.groupby('TP_FUNDO_CLASSE').agg({
    'CAPTC_DIA': 'mean',
    'RESG_DIA': 'mean'
})

print("Resultado - Média Diária de Captação e Resgate por Categoria:")
display(resumo_capt_resg)

print("\n💡 ANÁLISE PARA O GERENTE:")
print("- Comparar a entrada (captação) com a saída (resgate) revela o fluxo recente do mercado.")
print("- Orientação ao cliente: Se uma categoria específica apresentar resgates médios superiores às captações, alerte o cliente de que a classe está sofrendo pressão de resgate, o que pode forçar o gestor a vender ativos com desconto.")
```

```python
# PERGUNTA 9: Quais períodos apresentaram saldo positivo de captação líquida?

print("PERGUNTA 9: Quais dias do mês apresentaram saldo positivo de captação líquida?")


df['SALDO_LIQUIDO'] = df['CAPTC_DIA'] - df['RESG_DIA']
saldo_diario = df.groupby('DT_COMPTC')['SALDO_LIQUIDO'].sum()
dias_positivos = len(saldo_diario[saldo_diario > 0])
total_dias = df['DT_COMPTC'].nunique()

print(f"Resultado: {dias_positivos} de {total_dias} dias úteis no mês apresentaram saldo positivo de captação.")

print("\n💡 ANÁLISE PARA O GERENTE:")
print("- O comportamento do saldo diário mostra o otimismo ou pessimismo geral dos investidores no mês.")
print("- Orientação ao cliente: Em meses com mais dias de saldo negativo, oriente o cliente a não tomar decisões emotivas de resgate prévio, aproveitando o momento para compras fracionadas se os fundamentos do fundo continuarem sólidos.")
```

```python
# PERGUNTA 10: Quais categorias apresentaram o maior saldo líquido de captação (Captação - Resgate)?

print("PERGUNTA 10: Quais categorias de fundo apresentaram a maior captação líquida média?")


# Criação da variável de saldo líquido diário
df['SALDO_LIQUIDO'] = df['CAPTC_DIA'] - df['RESG_DIA']

# Agrupamento por categoria e cálculo da média do saldo líquido
saldo_por_classe = df.groupby('TP_FUNDO_CLASSE')['SALDO_LIQUIDO'].mean().sort_values(ascending=False).to_frame()
saldo_por_classe.columns = ['Saldo Líquido Médio (R$)']

print("Resultado - Captação Líquida Média por Categoria:")
display(saldo_por_classe)

print("\n💡 ANÁLISE PARA O GERENTE:")
print("- Categorias com saldo líquido positivo representam os segmentos do mercado em fase de expansão e preferência dos investidores.")
print("- Categorias com saldo líquido negativo indicam movimento de fuga de capital (resgates consolidados superiores às entradas).")
print("- Orientação ao cliente: Em momentos de elevada incerteza, direcione novos aportes para as classes com fluxo de entrada positivo, garantindo maior estabilidade operacional ao gestor e menor necessidade de liquidação forçada de ativos.")
```
