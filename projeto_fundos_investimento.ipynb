```python
import pandas as pd

# Carregar o dataset da CVM
df = pd.read_csv('dataset/inf_diario_fi_202609.csv', sep=';')

# Exibir as primeiras 5 linhas da tabela
df.head()
```

```python
# Verificar a quantidade de linhas e colunas
print("Formato do dataset (Linhas, Colunas):", df.shape)

# Exibir informações das colunas e tipos de dados
df.info()
```

```python
# Converter a coluna de data para o formato datetime
df['DT_COMPTC'] = pd.to_datetime(df['DT_COMPTC'])

# Verificar o intervalo de datas do arquivo
print("Data inicial:", df['DT_COMPTC'].min())
print("Data final:", df['DT_COMPTC'].max())

# Estatísticas descritivas do Patrimônio Líquido
df[['VL_PATRIM_LIQ', 'VL_QUOTA']].describe()
```

```python
# Formatar visualização do Pandas para mostrar decimais normais sem notação científica
pd.options.display.float_format = '{:,.2f}'.format

# Filtrar os dados do último dia disponível (28/09/2026)
ultimo_dia = df[df['DT_COMPTC'] == '2026-09-28']

# Listar os 10 maiores fundos por Patrimônio Líquido nesse dia
maiores_fundos = ultimo_dia.sort_values(by='VL_PATRIM_LIQ', ascending=False).head(10)

# Exibir CNPJ e Patrimônio Líquido dos maiores fundos
maiores_fundos[['CNPJ_FUNDO_CLASSE', 'VL_PATRIM_LIQ']]
```
